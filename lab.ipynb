```python
edges = pd.read_csv("edges.csv")
```

```python
edges.info()
```

```python
from collections import defaultdict

count = defaultdict(lambda: 0)
for edge in edges.type:
    count[edge] += 1
    
arr = [(item[1], item[0]) for item in count.items()]

sort_arr = sorted(arr, key= lambda x: x[0], reverse=True)
```

```python
import json

with open("relationship.json", "w", encoding="utf-8") as f:
    json.dump(sort_arr, f, indent=2, ensure_ascii=False)
```
